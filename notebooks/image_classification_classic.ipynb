# Recommended preprocessing

**Normalize every detected numeral before extracting features:**

- Convert to binary foreground/background.
- Crop tightly to the numeral bounding box.
- Preserve aspect ratio and center it on a fixed canvas, for example 32×32.
- Optionally deskew using image moments.
- Ensure consistent polarity: numeral pixels =1, background =0.

Normalization is often more important than the choice of classifier.

# Good Feature Groups for Classical Numeral Recognition

## 1. Projection Profiles

Compute the number of foreground pixels in each row and each column.

Horizontal projection:

\[
h_i = \sum_x I(x,i)
\]

Vertical projection:

\[
v_j = \sum_y I(j,y)
\]

Typical features:

- Horizontal projection profile
- Vertical projection profile
- Downsampled projections (e.g., 8 horizontal + 8 vertical bins)

**Advantages**

- Very simple to implement.
- Captures where the numeral is wide or narrow.
- Works well for fixed fonts.

---

## 2. Zoning (Grid Occupancy)

Divide the normalized image into a regular grid (e.g., \(4 \times 4\)) and compute the foreground density inside each cell.

\[
f_{r,c} =
\frac{1}{|Z_{r,c}|}
\sum_{(x,y)\in Z_{r,c}} I(x,y)
\]

Typical grids:

- \(4 \times 4\) → 16 features
- \(6 \times 6\) → 36 features

**Advantages**

- Preserves coarse spatial information.
- Easy to visualize.
- Robust to small deformations.

---

## 3. Image Moments

Compute normalized central moments or Hu invariant moments.

Central moments:

\[
\mu_{pq} =
\sum_x\sum_y
(x-\bar{x})^p(y-\bar{y})^qI(x,y)
\]

Possible features:

- Centroid
- Variance in x
- Variance in y
- Covariance
- Orientation
- Eccentricity
- Hu moments

**Advantages**

- Compact global shape descriptor.
- Translation and scale invariant.
- Widely used classical feature.

---

## 4. Contour Geometry

Extract the outer contour and compute geometric descriptors.

Typical features:

- Area
- Perimeter
- Bounding-box width
- Bounding-box height
- Aspect ratio
- Extent
- Circularity

\[
C=\frac{4\pi A}{P^2}
\]

- Convex hull area
- Solidity

\[
\text{Solidity}=
\frac{A}{A_{\text{convex hull}}}
\]

**Advantages**

- Describes the overall shape.
- Easy to explain.
- Useful for distinguishing thin and compact numerals.

---

## 5. Topological Features

Describe the topology of the numeral.

Typical features:

- Number of connected components
- Number of holes
- Hole area
- Hole centroid
- Relative hole position

Examples:

- 0 → one hole
- 6 → one lower hole
- 8 → two holes
- 9 → one upper hole

**Advantages**

- Extremely discriminative.
- Highly interpretable.
- Nearly invariant to small deformations.

---

## 6. Scan-Line Intersection Features

Place several horizontal and vertical scan lines across the image.

For each scan line compute:

- Number of foreground segments
- Number of transitions
- Length of foreground intervals
- Position of first and last intersection

**Advantages**

- Captures structural information.
- Simple implementation.
- Easy for students to understand.

---

## 7. Symmetry Features

Measure similarity between mirrored halves of the numeral.

Possible features:

- Vertical symmetry
- Horizontal symmetry
- Upper/lower symmetry
- Left/right symmetry

Example metric:

\[
S=
\frac1N
\sum|I(x,y)-I(W-1-x,y)|
\]

**Advantages**

- Helps distinguish symmetric and asymmetric digits.
- Very intuitive feature.

---

## 8. Distance Transform Features

Compute the Euclidean distance transform inside the foreground.

Possible features:

- Maximum distance
- Mean distance
- Variance
- Histogram of distances
- Regional averages

**Advantages**

- Encodes stroke thickness.
- Robust to small contour variations.

---

## 9. Histogram of Oriented Gradients (HOG)

Compute histograms of local edge orientations over small cells.

Typical parameters:

- 9 orientation bins
- 8×8 pixel cells
- 2×2 cell blocks

**Advantages**

- Excellent recognition performance.
- Robust to small shifts.
- Classical handcrafted feature descriptor.
- Often paired with a linear SVM.
